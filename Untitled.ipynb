```python
 # Import the libraries required for loading,
# analyzing, and visualizing the dataset.
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from huggingface_hub import HfApi #HfApi is a class
api = HfApi() #HfApi is a Python interface that lets our program communicate with the Hugging Face website.

files = api.list_repo_files(  #api call
    "codeparrot/apps",
    repo_type="dataset"
)

print(files)
```

```python
from huggingface_hub import hf_hub_download  #This(hf_hub_download) function is specifically designed to download a single file from a Hugging Face repository.

# Download train dataset
train_path = hf_hub_download(
    repo_id="codeparrot/apps",
    filename="train.jsonl",
    repo_type="dataset"
)

# Download test dataset
test_path = hf_hub_download(
    repo_id="codeparrot/apps",
    filename="test.jsonl",
    repo_type="dataset"
)

print("Train Path:", train_path)
print("Test Path :", test_path)
"""
Import hf_hub_download
        │
        ▼
Call hf_hub_download()
        │
        ▼
Repository = codeparrot/apps
        │
        ▼
Filename = train.jsonl
        │
        ▼
Internet Request(api call)
        │
        ▼
Hugging Face Server
        │
        ▼
Find train.jsonl
        │
        ▼
Download file
        │
        ▼
Save in local cache
        │
        ▼
Return local path
        │
        ▼
Store in train_path
"""
```

```python
import json

train_data = []

with open(train_path, "r") as f:
    for line in f:
        train_data.append(json.loads(line))

print("Training Problems:", len(train_data))
"""
open()

↓

Read each line

↓

Convert JSON to dictionary

↓

Append into list

↓

Now the dataset is in memory"
"""
```

```python
test_data = []

with open(test_path, "r") as f:
    for line in f:
        test_data.append(json.loads(line))

print("Testing Problems:", len(test_data))
```

```python
import json

invalid_rows = []

# Check whether every input_output string is valid JSON.

for i, io in enumerate(df["input_output"]):
    try:
        json.loads(io)
    except Exception:
        invalid_rows.append(i)

print("Invalid JSON rows:", len(invalid_rows))
```

```python
missing_keys = []

# Check whether every problem contains both 'inputs' and 'outputs'.

for i, io in enumerate(df["input_output"]):
    data_io = json.loads(io)

    if "inputs" not in data_io or "outputs" not in data_io:
        missing_keys.append(i)

print("Rows missing keys:", len(missing_keys))
```

```python
mismatch_rows = []

# Check whether the number of inputs equals the number of outputs.

for i, io in enumerate(df["input_output"]):
    data_io = json.loads(io)

    if len(data_io["inputs"]) != len(data_io["outputs"]):
        mismatch_rows.append(i)

print("Rows with mismatched test cases:", len(mismatch_rows))
# We verify that every input test case has exactly one corresponding expected output, ensuring the dataset is complete and suitable for training the model.
```

```python
# Count the number of test cases for every problem.

test_case_counts = []

for io in df["input_output"]:
    data_io = json.loads(io)
    test_case_counts.append(len(data_io["inputs"]))
```

```python
print("Minimum Test Cases :", min(test_case_counts))
print("Maximum Test Cases :", max(test_case_counts))
```
